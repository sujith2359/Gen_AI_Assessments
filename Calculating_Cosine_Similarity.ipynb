# Calculating Cosine Similarity Mathematically
This notebook demonstrates how to compute the cosine similarity between multi-dimensional vectors using three different approaches: a manual mathematical implementation, NumPy, and Scikit-Learn. We also explore edge cases, mathematical validation, and a practical application in Retrieval-Augmented Generation (RAG).

## 1. Define Vectors

```python
vector_A = [3, 1, 4, 2]
vector_B = [2, 5, 1, 3]

print("Vector A:", vector_A)
print("Vector B:", vector_B)
```

```python
import math

def calculate_norm(vec):
    sum_squares = sum(v ** 2 for v in vec)
    return math.sqrt(sum_squares)

norm_A = calculate_norm(vector_A)
norm_B = calculate_norm(vector_B)

print("Norm of A:", round(norm_A, 5))
print("Norm of B:", round(norm_B, 5))
```

```python
def manual_cosine(vec1, vec2):
    # 1. Dimensionality validation
    if len(vec1) != len(vec2):
        raise ValueError(f"Dimension mismatch error: {len(vec1)} vs {len(vec2)}")

    n1 = calculate_norm(vec1)
    n2 = calculate_norm(vec2)

    # 2. Zero vector check
    if n1 == 0 or n2 == 0:
        raise ValueError("Cosine similarity is undefined for zero vectors (magnitude is 0).")

    return calculate_dot(vec1, vec2) / (n1 * n2)

manual_sim = manual_cosine(vector_A, vector_B)
print("Manual Cosine Similarity:", round(manual_sim, 5))
```

```python
import numpy as np

arr_A = np.array(vector_A)
arr_B = np.array(vector_B)

numpy_dot = np.dot(arr_A, arr_B)
numpy_norm_A = np.linalg.norm(arr_A)
numpy_norm_B = np.linalg.norm(arr_B)

numpy_sim = numpy_dot / (numpy_norm_A * numpy_norm_B)
print("NumPy Cosine Similarity:", round(numpy_sim, 5))
```

```python
from sklearn.metrics.pairwise import cosine_similarity

# Reshape arrays to shape (1, -1) representing a single sample
reshape_A = arr_A.reshape(1, -1)
reshape_B = arr_B.reshape(1, -1)

sklearn_sim = cosine_similarity(reshape_A, reshape_B)[0][0]
print("Scikit-learn Cosine Similarity:", round(sklearn_sim, 5))
```

```python
all_close = np.isclose(manual_sim, numpy_sim) and np.isclose(numpy_sim, sklearn_sim)

print("Manual calculation matches NumPy? ", np.isclose(manual_sim, numpy_sim))
print("NumPy calculation matches Scikit-learn?", np.isclose(numpy_sim, sklearn_sim))
print("All three methods yield equivalent results:", all_close)
```

```python
test_pairs = {
    "Highly Similar Vectors": ([1, 2, 3], [1.1, 2.1, 3.1]),
    "Opposite-Direction Vectors": ([1, 2, 3], [-1, -2, -3]),
    "Orthogonal Vectors": ([1, 0, 0], [0, 1, 0]),
    "Random Vectors": ([4, -2, 7], [-1, 5, 2])
}

for scenario_name, (v1, v2) in test_pairs.items():
    score = manual_cosine(v1, v2)
    print(f"{scenario_name:28s} -> Cosine Score: {round(score, 5)}")
```

```python
# Let the dimensions represent occurrences of keywords: ['cloud', 'compute', 'database', 'ai', 'neural']
doc_embedding_1 = [5, 4, 1, 0, 0] # Tech document about cloud databases
doc_embedding_2 = [0, 1, 0, 6, 5] # Artificial intelligence paper

embedding_similarity = manual_cosine(doc_embedding_1, doc_embedding_2)
print("Document 1 Embedding:", doc_embedding_1)
print("Document 2 Embedding:", doc_embedding_2)
print("Semantic Cosine Similarity:", round(embedding_similarity, 5))
```
